```python

# 1. Определяем корень проекта 
try: 
    # Если код запущен как .py файл
    PROJECT_ROOT = Path(__file__).resolve().parent.parent 
except NameError:
    # Если код запущен в Jupyter Notebook
    # __file__ нет, поэтому берем текущую рабочую директорию (где запущен Jupyter)
    # и поднимаемся на уровень выше (к корню проекта) 
    PROJECT_ROOT = Path.cwd().parent
print(PROJECT_ROOT)

# Путь до файла pyproject.toml
file_path = Path(PROJECT_ROOT, 'pyproject.toml')

# Чтение файла pyproject.toml
with open(file_path, 'rb') as f:
    data = tomllib.load(f)

# Вывод содержимого файла
for key, value in data.items():
    print(f"{key}: {value}")
```
