```python
import os
import tomllib
from importlib import import_module
from pathlib import Path

# Указываем абсолютный путь до директории проекта
# Можно заменить на любой другой подходящий путь
PROJECT_ROOT = Path('/home/wirekit69/ai_engineer_preproject')

# Полный путь до файла pyproject.toml
pyproject_path = PROJECT_ROOT / 'pyproject.toml'

# Чтение файла pyproject.toml
with open(pyproject_path, 'rb') as f:
    data = tomllib.load(f)

# Получение списка зависимостей
dependencies = data.get('tool', {}).get('poetry', {}).get('dependencies', {})

for lib_name, version in dependencies.items():
    # Импорт модуля
    try:
        module = import_module(lib_name)
    except ImportError:
        print(f'Не удалось импортировать библиотеку: {lib_name}')
        continue

    # Проверка версии библиотеки
    try:
        print(f'{lib_name}: {module.__version__}')
    except AttributeError:
        print(f'Версия библиотеки {lib_name} недоступна')
```
